## Create a visual showing the relationship between horsepower and fuel economy.


```python
sns.relplot( data=mpg, x="mpg", y="horsepower")
```

<span style="color:green">Generally speaking, as the mileage of a car goes up, the horsepower goes down.</span>

## Create a table summarizing fuel economy by country of origin

```python
summary = (mpg.groupby("origin")["mpg"].agg(count = "count", 
    mean_mpg = "mean",
    median_mpg= "median",
    min_mpg = "min",
    max_mpg = "max").round(2).reset_index())
print(summary)
```

<span style="color:green">Japan has the best average fuel economy of each country, but the USA has the largest number of variables by far.</span>

## Create a visual showing how fuel economy has changed over time

```python
sns.lineplot(data=mpg, x="model_year", y="mpg",)
```

<span style="color:green">Over time, the average mpg of vehicles has risen. Recently, it fell a bit but has continued on the upward trend.</span>

## Create two other visuals or tables related to fuel economy.

```python
cylinders_to_mpg = (mpg.groupby(by = ["origin", "cylinders"])["mpg"].agg(mean_mpg = "mean"))
print(cylinders_to_mpg)
```

<span style="color:green">In conjunction with our previous information, we can now conclude that different numbers of cylinders may affect the mileage per country, as each country does not have the same cylinders in their vehicles. The USA can have an additional 2 cylinders, so it may attribute to the significantly lower mileage in the US.</span>

```python
sns.catplot(data = mpg, x="weight", y="mpg", hue="horsepower")
```

<span style="color:green">As the weight of the car increases, the mpg decreases. It's also interesting to note that as the weight increases, so does the horsepower. Weight may have some relationship to the horsepower. </span>
